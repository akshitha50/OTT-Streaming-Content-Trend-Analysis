```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing Value Percentage:")
print(missing_percentage.round(2))
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate records removed successfully!")
print("Updated Dataset Shape:", df.shape)
```

```python
columns_to_fill = ['director', 'cast', 'country', 'rating']

for col in columns_to_fill:
    df[col] = df[col].fillna('Unknown')

print("Missing values handled successfully!")
```

```python
print("Final Missing Values:")
print(df.isnull().sum())
```

```python
content_count = df['type'].value_counts()

print("Movies and TV Shows Count:")
print(content_count)
```

```python
content_percentage = df['type'].value_counts(normalize=True) * 100

print("Content Percentage:")
print(content_percentage.round(2))
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='type')

plt.title('Distribution of Movies and TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')

plt.show()
```

```python
plt.figure(figsize=(7,7))

content_count.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Movies vs TV Shows Percentage')
plt.ylabel('')

plt.show()
```

```python
print(df['listed_in'].head(10))
```

```python
genre_data = df['listed_in'].dropna().str.split(', ').explode()

print("Total Genre Entries:", len(genre_data))
```

```python
genre_counts = genre_data.value_counts()

print("Top 10 Popular Genres:")
print(genre_counts.head(10))
```

```python
plt.figure(figsize=(8,8))

genre_counts.head(10).plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 10 Genres Distribution')
plt.ylabel('')

plt.show()
```

```python
movies = df[df['type'] == 'Movie'].copy()

print("Total Movies:", len(movies))
```

```python
movies['duration_minutes'] = (
    movies['duration']
    .str.extract(r'(\d+)')
    .astype(float)
)

print(movies[['title', 'duration', 'duration_minutes']].head(10))
```

```python
average_duration = movies['duration_minutes'].mean()

print("Average Movie Duration:", round(average_duration, 2), "minutes")
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    movies['duration_minutes'].dropna(),
    bins=30,
    kde=True
)

plt.title('Distribution of Netflix Movie Durations')
plt.xlabel('Duration (Minutes)')
plt.ylabel('Number of Movies')

plt.show()
```

```python
plt.figure(figsize=(10,4))

sns.boxplot(x=movies['duration_minutes'])

plt.title('Movie Duration Box Plot')
plt.xlabel('Duration (Minutes)')

plt.show()
```
